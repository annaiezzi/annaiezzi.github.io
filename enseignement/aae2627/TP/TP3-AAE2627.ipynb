## 2. Division euclidienne de deux polynômes 

On a une notion de division euclidienne sur les polynômes comme sur les entiers :

Soit $A$ et $B$ deux polynômes non nuls à coefficients dans $\mathbb Q$. Alors il existe une unique paire de polynômes $Q,R$ tels que :

$$A = BQ + R, \quad \text{avec } R=0 \text{ ou } \deg(R) < \deg(B).$$

Comme avec les entiers, $Q$ est appelé *quotient* et $R$ *reste* de la division euclidienne de $A$ par $B$.

*Remarque :* en Sage, le degré du polynôme nul vaut $-1$, ce qui permet d'écrire simplement la condition $\deg(R) < \deg(B)$.

La division euclidienne de $A$ par $B$ peut se calculer ainsi, en notant $\operatorname{cd}(P)$ le coefficient dominant d'un polynôme $P$ :
- $Q \leftarrow 0$ et $R \leftarrow A$ ;
- tant que $\deg(R) \geq \deg(B)$ :
    - $T \leftarrow \dfrac{\operatorname{cd}(R)}{\operatorname{cd}(B)}\, Y^{\deg(R) - \deg(B)}$ ;
    - $Q \leftarrow Q + T$ et $R \leftarrow R - T B$ ;
- renvoyer $(Q, R)$.

_Remarque pour aller plus loin : La division euclidienne est un algorithme très pratique, donc on aimerait bien la définir sur des structures plus compliquées que les entiers. C'est le cas ici pour les polynômes, et plus généralement pour les anneaux dits_ euclidiens. 

_L'élément clé pour pouvoir définir cette division sur les polynômes est la fonction $\deg : \mathbb Q [Y] \rightarrow \mathbb N$. Cette fonction permet de définir une notion d'ordre (on veut que $\deg(R) <\deg(B)$), et, de plus, elle assure que l'algorithme termine (car $\deg(R)$ est strictement décroissante)._

_Le degré joue le rôle d'un_ stathme euclidien _, et son existence fournit énormément de propriétés sur un anneau._

_Plus généralement, sur un anneau $A$, un stathme euclidien est une fonction $s : A \rightarrow \mathbb N$ qui vérifie les propriétés suivantes :_

- $\forall a,b \in A, s(ab) \leqslant s(a) +s(b)$
- $ \forall a \in A, \forall b \in A \backslash \{0\}, \exists q,r \in A : a = qb +r \text{ et } s(r)<s(b)$ (division euclidienne)

## 3. PGCD de polynômes

Dans $\mathbb Q[Y]$, le *pgcd* de deux polynômes $A$ et $B$ non tous deux nuls est l'unique polynôme **unitaire** (c'est-à-dire de coefficient dominant égal à $1$) qui divise $A$ et $B$ et qui est de degré maximal parmi les diviseurs communs. C'est la convention utilisée par la fonction `gcd` de Sage.

_Remarque pour aller plus loin : Le PGCD est encore une notion très intéressante lorsqu'on fait de l'arithmétique sur un anneau. L'existence d'un PGCD découle d'une propriété beaucoup plus abstraite dont nous parlerons un peu en TD : la principalité._ 

_Un anneau $A$ est dit_ principal _lorsque ses idéaux sont tous de la forme $I = x A$ pour un élément $x \in A$. On démontrera en TD que cette propriété est vraie sur $\mathbb Z$. Sur $\mathbb Q[Y]$, la démonstration est un peu plus compliquée et nécessite d'utiliser les propriétés de la fonction degré. De fait cela découle de la proposition suivante :_
